```python
#Создадим функцию для перевода названий столбцов к snake_case
def to_snake_case(column_name):
#Удаляем лишние символы и приводим строку к нижнему регистру
    column_name = column_name.lower()
    
#Заменяем пробелы на _
    for char in ['  ']:
        if char in column_name:
            column_name = column_name.replace(char, '_')
            
    return column_name
```

```python
#Применим функцию к датафрейму
companies.columns = [to_snake_case(col) for col in companies.columns]
```

```python
#Выведем названия столбцов после преобразования
companies.columns
```

Столбцы `founded at`,`closed at`, `funded  at` необходимо перевести в тип даты и времени `datetime`.

В столбцах укажем медиану суммы финансирования и количество раундов финансирования.<br> Данные будем агрегировать по медиане, а не по среднему арифметическому, так как медиана учитывает центральное значение данных, в то время как среднее арифметическое может быть искажно из-за аномально высоких или низких значений сумм финансирования.

```python
#Составим сводную таблицу
raised_pivot =companies.pivot_table(index = 'funded_year', 
                             values = 'raised_amount',
                             aggfunc =['median', 'count'],
                                   fill_value=0)
raised_pivot
```

Отфильтруем сводную таблицу и оставим только года, где раудов финансирования больше 50.

```python
#Фильтруем сводную таблицы
filtered_pivot = raised_pivot[raised_pivot[('count', 'raised_amount')] > 50]
filtered_pivot
```

Построим график динамики типичного размера срдеств, которые стартапы получали в рамках одного раунда финанисрования.

```python
#Создадим фигуру для графики 12 на 8 дюймов
plt.figure(figsize=(12,8))

#Создадим линейный график
plt.plot(filtered_pivot.index, filtered_pivot[('median', 'raised_amount')], marker='o', color = 'blue')

#Добавляем название графика и подписи осей
plt.title('Динамика типичного размера средств финансирования')
plt.xlabel('Год финансирования')
plt.ylabel('Средний размер финансирования')

#Включаем сетку
plt.grid(True)

#Выводим график
plt.show()
```

 - малые компании, где 1 сотрудник;
 - средние предприятия, с количеством сотрудников от 2 до 20 сотрудников;

```python
#Обозначим размеры групп
total_employees['category'] = pd.cut(total_employees['id_count'], bins = [0,1,20,250], labels = ['малая', 'средняя', 'крупная'])
```

```python
#Проверим присвоение групп
total_employees.head()
```

Оценим долю сотрудников без образования в каждой из групп. Сотрудников без образования будем считать как сотрудники с пропусками в столбце `graduated_at` датасета `employees`. 

Объединим данные из `total_employees` и `employees` по `company_id` в датасет `no_education`.

```python
#Создадим датасет с категорией компаний
no_education = employees.merge(total_employees, how = 'inner', on = 'company_id')
```

```python
#Проверим и выведем первые 5 строк датасета
no_education.head()
```

Столбец с категорией компаний появился в датасете. Рассчитаем среднее количество сотрудников без образования в каждой категории. Для этого:
 - добавим новый столбец `not_graduated` к датасету `no_education`, который принимает булевое значение True/False по наличию образованию;
 - посчитаем количество сотрудников без образования в каждой компании и сохраним их в переменной `no_education_per_company` столбец `id_x`;
 - посчитаем долю сотрудников без образования в каждой компании, соединив датасет `no_education_per_company` и `total_employees`, результат помести в столбец `share`;
 - сгруппируем данные по категориям компаний и посчитаем среднюю долю сотрудников без образования.

```python
#Добавим столбец с признаком наличия/отсутствия образования
no_education['not_garduated'] = no_education['graduated_at'].isna()
```

```python
#Посчитаем количество сотрудников без образования в каждой компании
no_education_per_company = no_education.query('not_garduated == True').groupby(['company_id'])['id'].count().reset_index(name = 'id_x')

#Выведем полученные данные
no_education_per_company.head()
```

```python
#Соединим датасеты
no_education_per_company = total_employees.merge(no_education_per_company, how  = 'left', on = 'company_id')

#Выведм к просмотру первые 5 строк получишегося датасета
no_education_per_company.head()
```

```python
#Для расчета доли сначала в столбце id_x заменим пропуски на нули
no_education_per_company['id_x'] = no_education_per_company['id_x'].fillna(0)
```

```python
#Посчитаем долю сотрудинков без образования и сохраним в столбце share
no_education_per_company['share'] = (no_education_per_company['id_x'] / no_education_per_company['id_count'])*100

#Выведм первые пять строк датасета с новым столбцом
no_education_per_company.head()
```

```python
#Посчитаем среднюю долю людей без образования по категориям
category_education = no_education_per_company.groupby('category')['share'].mean().round()

#Выведем полученные данные
category_education
```

Рассчитав среднюю долю для сотрудников без образования, можно сделать следующие выводы:

 - в **средних** компаниях, где работает от 2 до 20 сотрудников доля сотрудников без образования **62%**;
 - в **крупных компаниях**, где работает больше 20 сотрудников, доля сотрудников без образования **67%**.

Самая высокая доля сотрудников без образования в малых компаниях, где работает один сотрудник. Это может быть связано с тем, что сотрудник он же владелец компании еще не закончил обучение, но уже решил создать свою компанию.

### 2.3. Объединять или не объединять — вот в чём вопрос

В нескольких датасетах присутствует столбец `network_username`. Необходимо проанализировать дублируется ли для разных датасетов информация в столбцах с таким названием, и если да, то насколько часто.

Также необходимо оценить можно ли использовать столбцы с именем `network_username` для объединения данных.

Столбец `network_username` встречается в датасете `companies` и в датасете `people`, но в датасетах у этого столбца разное описание.<br> В датасете `companies` этот столбец означает ник компании в соцсетях, а в датасете `people` ник сотрудника в социальных сетях. Но, так как в предудыщей задаче мы выяснили, что компаний с 1 сотрудником больше 75%, то есть вероятность что данные в этих столбцах дублируются.

Для того, чтобы проверить столбец на дубликаты мы сделаем следующее:
 - проанализируем количество строк столбце в каждом датасете;
 - посчитаем количество уникальных значений в этих столбцах;
 - присоединим строки датасета `people` к датасету `companies`. Сохраним данные в переменной `mix_df`;
 - удалим все пропуски в данных, так как они могут исказть дубликаты, если будут пропуски NaN-Nan. Новый датасет сохраним в переменной `mixed`.
 - посчитаем количество дубликатов в новом датасете, процет дубликатов;
 - выведем частоту повторяющихся значений.

```python
#Считаем количество строк в датасетах companies и people
companies_str= companies['network_username'].shape[0]
people_str = people['network_username'].shape[0]

print(f'В датасете companies в столбце network_username {companies_str} строк')
print(f'В датасете people в столбце network_username {people_str} строк')
```

```python
#Считаем количество уникальных значений в датасетах
uniq_companies = companies['network_username'].nunique()
people_unique = people['network_username'].nunique()

print(f'В датасете companies в столбце network_username {uniq_companies} уникальных значений')
print(f'В датасете people в столбце network_username {people_unique} уникальных значений')
```

```python
#Создаем копию датасета companies
combined = companies[['network_username']].copy()

#Добавляем к нему строки из people
mix_df = combined.append(people[['network_username']], ignore_index=True)

#В новом датасете удалим пропуски
mixed = mix_df.dropna()
```

```python
#Проверяем дубликаты
duplicates_count = mixed['network_username'].duplicated().sum()

#Считаем процент дубликатов
duplicates_percentage = (duplicates_count / len(mixed)) * 100

print(f"Количество дубликатов: {duplicates_count}")
print(f"Процент дубликатов: {duplicates_percentage:.2f}%")
```

```python
#Считаем частоту вхождений значений
duplicates_freq = mixed['network_username'].value_counts()

#Выведем получивштеся пять строк частоты вхождений. Первыц столбец - значений, второй - частота.
duplicates_freq.head()
```

В столбце `network_username` информация дублируется, но не часто. Всего найдено 18608 совпадающих значений или 13.85% дубликатов. Использовать столбец `network_username` для объединения датасетов не рекомендуется, так как будет большое количество пропусков или будет утеряна большая часть значений, в зависимости от типа объединения.

Закрытые стартапы не успевают привлечь много финансирования и пройти много этапов финансирования.<br> Для статуса `ipo`, когда компания выходит на фондовый рынок со своими акциями, обычно проходит 2-3 раунда финансирования, типичное значение - 2, максимальное - 13. Это значит, что успешные компании, которые доходят до IPO обычно требуют большего количества раундов финансирования.<br> Для статуса `operating` в среднем проходит - 1-2 раунда финансирования, типичное значение - 1, максимальное - 15.

Визулизируем полученные данные и построим график, который покажет сколько в среднем раундов финансирования проходит для компании в каждом из статусов.
