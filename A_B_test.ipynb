## 2. Проверка гипотезы о среднем времени активности пользователей

Была выдвинута гипотеза, что пользователи из Санкт-Петербурга проводят в среднем больше времени за чтением и прослушиванием книг в приложении, чем пользователи из Москвы. 
- Нулевая гипотеза H₀: Средняя активность пользователей в часах в двух группах (Москва и Санкт-Петербург) не различается.

- Альтернативная гипотеза H₁: Средняя активность пользователей в Санкт-Петербурге больше, и это различие статистически значимо.

Так как данные распределены ассиметрично, а не нормально, то для проверки гипотезы используем тест Манна-Уитни. Уровень достоверности теста установим 95%.

```python
#Импортируем библиотеки для проведения теста
from scipy.stats import mannwhitneyu
import numpy as np
```

```python
#Устанавливаем уровень значимости
alpha = 0.05

mw_test = mannwhitneyu( 
    moscow['hours'],  #Выборка значений метрики для Москвы
    spb['hours'],  #Выборка значений метрики для Санкт-Петербурга
    alternative='less' #Выбираем альтернативную гипотезу moscow < spb
)

if mw_test.pvalue > alpha:
    print(f'pvalue={mw_test.pvalue} > {alpha}')
    print('Нулевая гипотеза находит подтверждение!  Средняя активность пользователей в часах в двух группах (Москва и Санкт-Петербург) не различается.')
else:
    print(f'pvalue={mw_test.pvalue} < {alpha}')
    print('Нулевая гипотеза не находит подтверждения! Средняя активность пользователей в Санкт-Петербурге больше, и это различие статистически значимо.')
```

## 3. Аналитическая записка

Проведен тест Манна-Уитни для проверки гипотезы о том, что средняя активность пользователей в часах в Санкт-Петербурге выше, чем в Москве. Уровень статистической значимости выбран 5%, стандартное значение для анализа данных, с которым мы готовы допускать ошибки первого рода в эксперименте.

Тест показал результат **p_value равный 0.27**, что больше уровня значимости. Это означает, что вероятность случайного возникновения наблюдаемого различия велика. Соответственно, оснований для отказа от нулевой гипотезы нет. То есть имеющиеся данные не подтверждают утверждение, что активность пользователей в Санкт-Петербурге выше, чем в Москве.

Можно сказать, что **между LTV и активностью пользователей в Москве и Санкт-Петербурге отсутствует корреляция**. Более высокий LTV в Москве связан с большим количеством пользователей, чем в Санкт-Петербурге.
Также пользователи могут покупать книги, но не слушать/читать их.

Оцените достаточность выборки для получения статистически значимых результатов A/B-теста. Заданные параметры:

- базовый показатель конверсии — 30%,

- мощность теста — 80%,

- достоверность теста — 95%.

Также для корректности проведения теста проанализируем достаточность выборки для получения статистически значимых результатов A/B теста. 

Параметры для проверки:
 - базовый показатель конверсии 30%
 - ожидаем эффект +3% к базовой конверсии
 - мощность теста 80%
 - достоверность теста 95%.

```python
import statsmodels.stats.api as sms

#Задаём параметры ожидаемого эффекта
effect_size = sms.proportion_effectsize(0.3, 0.33)

#Рассчитываем размер выборки
required_n = sms.NormalIndPower().solve_power(effect_size, power=0.8, alpha=0.05, ratio=1)

print(f'Необходимый размер выборки для каждой группы: {int(required_n)}')
```

Необходимый размер выборки для каждой группы должен быть не менее 3761 пользователя. Проверим размер выборки в данных о проведенном тесте в датасете `df_test`.

```python
print(f"Размер выборки группы А: {df_test[df_test['group'] == 'A']['user_id'].nunique()}")
print(f"Размер выборки группы B: {df_test[df_test['group'] == 'B']['user_id'].nunique()}")
```

Размеры выборки групп превышают рассчитанное значение в 3761 пользователя, поэтому можем считать, что результат теста будет надежным.

Для проведения теста рассчитаем общее количество пользователей (`total_users`), количество пользователей совершивших покупку (`buyers`) и конверсию по группам (`conversion`).

```python
#Общее количество пользователей
total_users = df_test.groupby('group')['user_id'].nunique()
total_users
```

```python
#Количество пользователей, совершивших покупку
buyers = df_test[df_test['event_name']=='purchase'].groupby('group')['user_id'].nunique()
buyers
```

```python
#Конверсия в покупку между группами ползьователей
conversion = buyers/total_users
conversion.round(2)
```

Предварительно, оценив полученные результаты обшего количества пользователей, покупателей и конверсии, можно сказать, что в тестовой группе конверсия в покупку выросла на 2% пункта. Проверим изменение конверсии далее с помощью статистического теста.

## 4. A/B тест и оценка конверсии:

Для оценки изменения конверсии был выбран z-тест пропорций, так как метрика, которую проверяем является долевой (доля пользователей, совершивших покупку в течение семи дней после регистрации).

```python
from statsmodels.stats.proportion import proportions_ztest

#Рассчитаем размер группы A
n_a = total_users['A']

#Рассчитаем размер группы B
n_b = total_users['B']

#Рассчитаем количество покупок группы A
m_a = buyers['A']

#Рассчитаем количество покупок группы B
m_b = buyers['B']


#Рассчитаем доли покупок в каждой группе
p_a, p_b = m_a/n_a, m_b/n_b

print(f'Размер группы А - {n_a}, размер группы B - {n_b}, покупок в группе А - {m_a}, покупок в группе B - {m_b}')

#Проверяем предпосылку о достаточном количестве данных
if (p_a*n_a > 10)and((1-p_a)*n_a > 10)and(p_b*n_b > 10)and((1-p_b)*n_b > 10): 
    print('Предпосылка о достаточном количестве данных выполняется!')
else:
    print('Предпосылка о достаточном количестве данных НЕ выполняется!')

    
#Устанавливаем уровень значимости    
alpha = 0.05

#Проводим Z-тест пропорций
stat_ztest, p_value_ztest = proportions_ztest(
    [m_a, m_b],
    [n_a,n_b],
    alternative='smaller' #Альтернативная гипотеза p_a < p_b
)


#Выводим значение p_value
print(f'pvalue={p_value_ztest}')
print(f'stat_ztest={stat_ztest}')   

#Формируем итоговый вывод
if p_value_ztest > alpha:
    print('Конверсия в покупку не увеличилась')
else:
    print('Конверсия в покупку увеличилась') 
```

На основании проведенного теста можно сделать следующие выводы:
 - данных для оценки результатов теста достаточно, количество пользователей в группах больше рассчитанного минимального значения выборки;
 - получеенное значение p_value меньше установленного уровня значимости теста, а значит можно сказать, что нулевая гипотеза о том, что конверсия не увеличилась, отвергается;
 - результаты показали увеличение конверсии в покупку в группе B на 2,51% больше, чем в группе A;
 - новый интерфейс действительно можно считать более удобным для пользователей, он увеличивает конверсию в покупку на ожидаемые значения. Можно рассмотреть предложение о раскатывании нового интерфейса на всех пользователей.
